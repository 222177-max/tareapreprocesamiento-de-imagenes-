## Importación de librerías
En esta actividad importaremos las bibliotecas necesarias para trabajar con imágenes. OpenCV permitirá procesarlas, NumPy realizará operaciones numéricas y Matplotlib permitirá visualizar los resultados

```python
# Importar OpenCV para el procesamiento de imágenes
import cv2

# Importar NumPy para realizar operaciones matemáticas
import numpy as np

# Importar Matplotlib para mostrar las imágenes
import matplotlib.pyplot as plt

# Mostrar las imágenes dentro del cuaderno Jupyter
%matplotlib inline

print("Librerías importadas correctamente")
```

 # Lectura de la imagen original en RGB
 En esta actividad cargaremos la fotografía del automóvil desde nuestra carpeta de trabajo. OpenCV lee inicialmente las imágenes en formato BGR, por lo que realizaremos una conversión a RGB para obtener el orden correcto de los colores.

```python
# Leer la imagen del automóvil desde la carpeta
imagen = cv2.imread("automovil.jpg")

# Verificar que la imagen se haya cargado correctamente
if imagen is None:
    raise FileNotFoundError("No se encontró la imagen automovil.jpg")

# Convertir la imagen de BGR a RGB
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

# Mostrar las dimensiones de la imagen
print("Dimensiones de la imagen:", imagen_rgb.shape)

# Mostrar la imagen original en RGB
plt.figure(figsize=(8, 5))
plt.imshow(imagen_rgb)
plt.title("Imagen Original en RGB")
plt.axis("off")
plt.show()
```

### interpretacion
Se cargó correctamente la fotografía del automóvil y se convirtió del formato BGR al formato RGB. Finalmente, se visualizó la imagen original, que servirá como referencia para realizar los siguientes procedimientos de preprocesamiento.

# Redimensionamiento de la imagen a 64 × 64 píxeles
En esta actividad cambiaremos las dimensiones de la imagen original a 64 píxeles de ancho y 64 píxeles de alto. Esto permite trabajar con imágenes de un tamaño uniforme y facilita su procesamiento posterior

```python
# Redimensionar la imagen RGB a 64 píxeles de ancho y alto
imagen_64 = cv2.resize(imagen_rgb, (64, 64))

# Mostrar las dimensiones de la nueva imagen
print("Nuevas dimensiones:", imagen_64.shape)

# Crear una figura para comparar las imágenes
plt.figure(figsize=(10, 5))

# Mostrar la imagen original
plt.subplot(1, 2, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen Original")
plt.axis("off")

# Mostrar la imagen redimensionada
plt.subplot(1, 2, 2)
plt.imshow(imagen_64)
plt.title("Imagen 64 x 64 píxeles")
plt.axis("off")

# Mostrar los resultados
plt.show()
```

### interpretacion
 Se realizó el redimensionamiento de la imagen original a 64 × 64 píxeles. Se observa que la imagen conserva su contenido, pero presenta una reducción considerable de resolución y algunos detalles pueden perderse debido al menor número de píxeles

# Conversión de RGB a escala de grises
En esta actividad convertiremos la imagen RGB de 64 × 64 píxeles a escala de grises. Este procedimiento elimina la información de color y conserva los niveles de intensidad, facilitando el análisis de las imágenes

```python
# Convertir la imagen RGB a escala de grises
imagen_gris = cv2.cvtColor(imagen_64, cv2.COLOR_RGB2GRAY)

# Crear una figura para comparar los resultados
plt.figure(figsize=(10, 5))

# Mostrar la imagen original RGB
plt.subplot(1, 2, 1)
plt.imshow(imagen_64)
plt.title("Imagen   RGB 64 x 64")
plt.axis("off")

# Mostrar la imagen en escala de grises
plt.subplot(1, 2, 2)
plt.imshow(imagen_gris, cmap="gray")
plt.title("Imagen en Escala de Grises")
plt.axis("off")

# Mostrar los resultados
plt.show()
```

### Interpretación:

Se convirtió correctamente la imagen RGB a escala de grises. Se observa que desaparecen los colores originales del automóvil y se mantienen diferentes tonalidades entre blanco, negro y gris, permitiendo identificar las formas y detalles de la imagen.

# Conversión de RGB al espacio de color HSV
En esta actividad transformaremos la imagen RGB al espacio de color HSV, que representa los colores mediante tres componentes: tono (Hue), saturación (Saturation) y valor (Value

```python
# Convertir la imagen original RGB al espacio HSV
imagen_hsv = cv2.cvtColor(imagen_rgb, cv2.COLOR_RGB2HSV)

# Separar los tres componentes del espacio HSV
H, S, V = cv2.split(imagen_hsv)

# Mostrar la imagen original y los componentes HSV
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen Original RGB")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(H, cmap="gray")
plt.title("Componente H - Tono")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(S, cmap="gray")
plt.title("Componente S - Saturación")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(V, cmap="gray")
plt.title("Componente V - Brillo")
plt.axis("off")

plt.tight_layout()
plt.show()
```

### interpretacion
Se realizó la conversión de la imagen original RGB al espacio de color HSV utilizando OpenCV. Este espacio representa la información del color mediante tres componentes: tono (H), saturación (S) y valor (V). La separación de estos componentes permite analizar individualmente las características de color y luminosidad de la imagen.

# Ajustar el brillo y contraste de la imagen en escala de grises
modificar el brillo y el contraste de la imagen para observar cómo cambian sus niveles de iluminación.

```python
# AJUSTE DE BRILLO Y CONTRASTE


# Convertir la imagen original RGB a escala de grises
imagen_gris = cv2.cvtColor(imagen_rgb, cv2.COLOR_RGB2GRAY)

# Definir los valores de contraste y brillo
contraste = 1.5
brillo = 40

# Aplicar el ajuste de brillo y contraste
imagen_ajustada = cv2.convertScaleAbs(
    imagen_gris,
    alpha=contraste,
    beta=brillo
)

# Crear una figura para mostrar las imágenes
plt.figure(figsize=(15, 6))

# Mostrar la imagen original en RGB
plt.subplot(1, 3, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen Original RGB")
plt.axis("off")

# Mostrar la imagen original en escala de grises
plt.subplot(1, 3, 2)
plt.imshow(imagen_gris, cmap="gray")
plt.title("Imagen en Escala de Grises")
plt.axis("off")

# Mostrar la imagen con brillo y contraste ajustados
plt.subplot(1, 3, 3)
plt.imshow(imagen_ajustada, cmap="gray")
plt.title("Imagen con Brillo y Contraste")
plt.axis("off")

# Ajustar la distribución de las imágenes
plt.tight_layout()

# Mostrar los resultados
plt.show()
```

### Interpretación 
Se realizó el ajuste de brillo y contraste de la imagen en escala de grises utilizando la función cv2.convertScaleAbs() de OpenCV. Se estableció un valor de contraste de 1.5 y un valor de brillo de 40. Como resultado, se obtuvo una imagen más iluminada y con mayor diferencia entre las zonas claras y oscuras.

Explicación de los parámetros:
- alpha = 1.5: controla el contraste de la imagen.
- beta = 40: controla el brillo de la imagen.

# Detección de bordes utilizando SOBEL
Objetivo: identificar los bordes y contornos del automóvil mediante el operador Sobel, que permite detectar cambios de intensidad en la imagen.

```python

# DETECCIÓN DE BORDES MEDIANTE SOBEL

# Convertir la imagen RGB de 64x64 a escala de grises
imagen_gris = cv2.cvtColor(imagen_64, cv2.COLOR_RGB2GRAY)

# Detectar los bordes en dirección horizontal (X)
sobel_x = cv2.Sobel(
    imagen_gris,
    cv2.CV_64F,
    1, 0,
    ksize=3
)

# Detectar los bordes en dirección vertical (Y)
sobel_y = cv2.Sobel(
    imagen_gris,
    cv2.CV_64F,
    0, 1,
    ksize=3
)

# Combinar los resultados de ambas direcciones
sobel = cv2.magnitude(sobel_x, sobel_y)

# Normalizar los valores para visualizarlos entre 0 y 255
sobel = cv2.normalize(
    sobel,
    None,
    0,
    255,
    cv2.NORM_MINMAX
)

# Convertir el resultado a formato de 8 bits
sobel = np.uint8(sobel)


# MOSTRAR LOS RESULTADOS

plt.figure(figsize=(15, 6))

# Mostrar la imagen original RGB
plt.subplot(1, 3, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen Original RGB")
plt.axis("off")

# Mostrar la imagen en escala de grises
plt.subplot(1, 3, 2)
plt.imshow(imagen_gris, cmap="gray")
plt.title("Imagen en Escala de Grises")
plt.axis("off")

# Mostrar la imagen procesada con Sobel
plt.subplot(1, 3, 3)
plt.imshow(sobel, cmap="gray")
plt.title("Detección de Bordes - SOBEL")
plt.axis("off")

# Ajustar la presentación
plt.tight_layout()

# Mostrar las imágenes
plt.show()
```

### interpretacion
Se aplicó el operador Sobel para detectar los bordes de la imagen en las direcciones horizontal y vertical. Se combinaron ambos resultados para resaltar los contornos del automóvil y otros elementos presentes en la fotografía. Esta técnica permite identificar los cambios de intensidad y las características estructurales de la imagen.

# Detección de bordes utilizando CANNY

```python
# DETECCIÓN DE BORDES MEDIANTE CANNY

# Convertir la imagen original RGB a escala de grises
imagen_gris = cv2.cvtColor(imagen_64, cv2.COLOR_RGB2GRAY)

# Aplicar un filtro Gaussiano para reducir el ruido
imagen_suavizada = cv2.GaussianBlur(imagen_gris, (5, 5), 0)

# Aplicar el detector de bordes Canny
canny = cv2.Canny(imagen_suavizada, 80, 160)

# MOSTRAR LOS RESULTADOS

plt.figure(figsize=(15, 6))

# Mostrar la imagen original RGB
plt.subplot(1, 3, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen Original RGB")
plt.axis("off")

# Mostrar la imagen en escala de grises
plt.subplot(1, 3, 2)
plt.imshow(imagen_gris, cmap="gray")
plt.title("Imagen en Escala de Grises")
plt.axis("off")

# Mostrar la imagen procesada con Canny
plt.subplot(1, 3, 3)
plt.imshow(canny, cmap="gray")
plt.title("Detección de Bordes - CANNY")
plt.axis("off")

plt.tight_layout()
plt.show()
```

### Interpretación 
Se aplicó el algoritmo Canny para detectar los bordes de la imagen. Previamente, se utilizó un filtro Gaussiano para reducir el ruido. Posteriormente, se aplicaron umbrales de 80 y 160 para identificar los bordes más relevantes del automóvil. El resultado permite observar los contornos y las características principales de la imagen.

# Aumento de datos (Data Augmentation)

## Volteo vertical de la imagen
invertir la imagen verticalmente, haciendo que la parte superior quede abajo y la inferior arriba.

```python

# AUMENTO DE DATOS: VOLTEO VERTICAL


# Aplicar volteo vertical a la imagen original
imagen_volteada = cv2.flip(imagen_rgb, 0)


# MOSTRAR LOS RESULTADOS


plt.figure(figsize=(14, 7))

# Mostrar la imagen original sin reducir
plt.subplot(1, 2, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen Original")
plt.axis("off")

# Mostrar la imagen con volteo vertical
plt.subplot(1, 2, 2)
plt.imshow(imagen_volteada)
plt.title("Imagen con Volteo Vertical")
plt.axis("off")

# Ajustar la presentación
plt.tight_layout()
plt.show()
```

### interpretacion
Se aplicó la técnica de aumento de datos mediante el volteo vertical utilizando la función cv2.flip() de OpenCV. La imagen original fue invertida verticalmente, haciendo que la parte superior pase a la parte inferior y viceversa, conservando su resolución y calidad original

## Rotación de la imagen a 45 grados
girar la imagen del automóvil 45 grados para modificar su orientación y generar una nueva versión de la fotografía

```python
# AUMENTO DE DATOS: ROTACIÓN DE 45 GRADOS


# Obtener las dimensiones de la imagen original
alto, ancho = imagen_rgb.shape[:2]

# Calcular el centro de la imagen
centro = (ancho // 2, alto // 2)

# Crear la matriz de rotación de 45 grados
matriz_rotacion = cv2.getRotationMatrix2D(centro, 45, 1.0)

# Calcular el tamaño necesario para evitar recortar la imagen
nuevo_ancho = int(alto * abs(np.sin(np.radians(45))) +
                  ancho * abs(np.cos(np.radians(45))))
nuevo_alto = int(alto * abs(np.cos(np.radians(45))) +
                 ancho * abs(np.sin(np.radians(45))))

# Ajustar la posición de la imagen rotada
matriz_rotacion[0, 2] += (nuevo_ancho / 2) - centro[0]
matriz_rotacion[1, 2] += (nuevo_alto / 2) - centro[1]

# Aplicar la rotación a la imagen original
imagen_rotada = cv2.warpAffine(
    imagen_rgb,
    matriz_rotacion,
    (nuevo_ancho, nuevo_alto)
)

# MOSTRAR LOS RESULTADOS
plt.figure(figsize=(14, 7))

# Mostrar la imagen original
plt.subplot(1, 2, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen Original")
plt.axis("off")

# Mostrar la imagen rotada 45 grados
plt.subplot(1, 2, 2)
plt.imshow(imagen_rotada)
plt.title("Imagen Rotada 45 Grados")
plt.axis("off")

# Ajustar la presentación
plt.tight_layout()
plt.show()
```

### Interpretación 
Se aplicó la técnica de aumento de datos mediante la rotación de 45 grados utilizando las funciones cv2.getRotationMatrix2D() y cv2.warpAffine() de OpenCV. Esta transformación modifica la orientación de la fotografía, conservando la mayor parte de su contenido y generando una nueva variación de la imagen original.

## Traslación horizontal y vertical de la imagen
desplazar la imagen hacia la derecha y hacia abajo, generando una nueva posición del automóvil sin modificar su resolución original

```python

# AUMENTO DE DATOS: TRASLACIÓN


# Obtener las dimensiones de la imagen original
alto, ancho = imagen_rgb.shape[:2]

# Definir los desplazamientos
desplazamiento_x = 100  # Desplazamiento horizontal
desplazamiento_y = 60   # Desplazamiento vertical

# Crear la matriz de traslación
matriz_traslacion = np.float32([
    [1, 0, desplazamiento_x],
    [0, 1, desplazamiento_y]
])

# Aplicar la traslación a la imagen original
imagen_trasladada = cv2.warpAffine(
    imagen_rgb,
    matriz_traslacion,
    (ancho, alto)
)

# ------------------------------------------
# MOSTRAR LOS RESULTADOS
# ------------------------------------------

plt.figure(figsize=(14, 7))

# Mostrar la imagen original
plt.subplot(1, 2, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen Original")
plt.axis("off")

# Mostrar la imagen trasladada
plt.subplot(1, 2, 2)
plt.imshow(imagen_trasladada)
plt.title("Imagen con Traslación")
plt.axis("off")

# Ajustar la presentación
plt.tight_layout()
plt.show()
```

### Interpretación 
Se aplicó la técnica de aumento de datos mediante la traslación horizontal y vertical utilizando la función cv2.warpAffine() de OpenCV. La imagen fue desplazada 100 píxeles hacia la derecha y 60 píxeles hacia abajo, conservando sus dimensiones originales. Esta transformación permite generar nuevas variaciones de posición de una misma imagen

# Zoom de la imagen (acercamiento)
Objetivo: aumentar el tamaño visual del automóvil mediante un acercamiento (zoom), utilizando la fotografía original sin reducir su resolución.

```python

# AUMENTO DE DATOS: ZOOM


# Obtener las dimensiones de la imagen original
alto, ancho = imagen_rgb.shape[:2]

# Definir el factor de zoom
factor_zoom = 1.5

# Aumentar el tamaño de la imagen
imagen_ampliada = cv2.resize(
    imagen_rgb,
    None,
    fx=factor_zoom,
    fy=factor_zoom,
    interpolation=cv2.INTER_LINEAR
)

# Obtener las dimensiones de la imagen ampliada
alto_zoom, ancho_zoom = imagen_ampliada.shape[:2]

# Calcular el centro para realizar el recorte
inicio_x = (ancho_zoom - ancho) // 2
inicio_y = (alto_zoom - alto) // 2

# Recortar el centro para conservar las dimensiones originales
imagen_zoom = imagen_ampliada[
    inicio_y:inicio_y + alto,
    inicio_x:inicio_x + ancho
]


# MOSTRAR LOS RESULTADOS


plt.figure(figsize=(14, 7))

# Mostrar la imagen original
plt.subplot(1, 2, 1)
plt.imshow(imagen_rgb)
plt.title("Imagen Original")
plt.axis("off")

# Mostrar la imagen con zoom
plt.subplot(1, 2, 2)
plt.imshow(imagen_zoom)
plt.title("Imagen con Zoom (1.5x)")
plt.axis("off")

# Ajustar la presentación
plt.tight_layout()
plt.show()
```

### Interpretación 
Se aplicó la técnica de aumento de datos mediante zoom utilizando la función cv2.resize() de OpenCV. Se incrementó el tamaño de la imagen con un factor de 1.5 y posteriormente se realizó un recorte central para conservar las dimensiones originales. Esta transformación permite generar una nueva variación de la fotografía mediante un acercamiento a los objetos.

Parámetro utilizado:

- factor_zoom = 1.5: aumenta la imagen un 50 %.
